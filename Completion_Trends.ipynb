**COMPLETION TRENDS**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Cleaned_Preprocessed_Dataset_Week1.csv')
```

**COMPLETION RATE OVER TIME (STABILITY)**

This shows hours that produced more serious applicants and
If there are certain times linked with lower follow-up








```python
#Convert to DateTime
df['Entry created at'] = pd.to_datetime(df['Entry created at'])

#Create Completed flag
df['Completed'] = df['Status Description'] == 'Applied'

#convert hour into a new column
df['Signup_Hour'] = df['Entry created at'].dt.hour

#Group by hour and calculate mean completion rate
completion_rate_hourly = df.groupby('Signup_Hour')
['Completed']

#Plot
fig, ax = plt.subplots(figsize=(12,6))
completion_rate_hourly.plot(ax=ax, color='green', marker='o')
ax.set_title("Completion Rate by Hour of Signup")
ax.set_xlabel("Hour of Day")
ax.set_ylabel("Completion Rate")
ax.grid(True)
plt.tight_layout()
plt.show()
```

**COMPLETION TIME VARIATION (SPEED RO APPLY)**

This shows if successful applicants apply faster, if non-completers delays during application and if there are extreme outliers.

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Completed',
            y='Days_to_Apply',
            hue = 'Completed',
            legend = False, data=df,
            palette='Set2')
plt.title("Days to Apply: Completed vs Not Completed")
plt.xlabel("Completed Application")
plt.ylabel("Days to Apply")
plt.show()
```

**ENGAGEMENT VS COMPLETION**

This links engagement to completion likelihood

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Completed',
            y= 'Engagement_Score',
            hue = 'Completed',
            legend = False, data=df,
            palette='coolwarm')
plt.title("Engagement Score: Completed vs Not Completed")
plt.xlabel("Completed Application")
plt.ylabel("Engagement Score")
plt.show()
```
