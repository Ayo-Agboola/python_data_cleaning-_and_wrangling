```python
import pandas as pd
```

```python
df = pd.read_csv('Cleaned_Preprocessed_Dataset_Week1.csv')
```

```python
df.columns
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe(include='all')
```

Signups per year

```python
df['Signup_Year'].value_counts().sort_index()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure()
sns.countplot(x='Signup_Year', data=df, color='skyblue')
plt.title("Signups per Year")
plt.xlabel("Year")
plt.ylabel("Number of Signups")
plt.show()
plt.close()
```

Signups per Month

```python
df['Signup_Month_Name'].value_counts()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x='Signup_Month_Name', data=df,
order=df['Signup_Month_Name'].value_counts().index,
              hue='Signup_Month_Name',
              palette= 'viridis')
plt.title("Signups by Month")
plt.xticks(rotation=45)
plt.show()
plt.close()
```

Signups by Day of Week

```python
df['Signup_Day_Name'].value_counts()
```

```python
plt.figure()
sns.countplot(x='Signup_Day_Name', data=df,
              order=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'])
plt.title("Signups by Day of Week")
plt.show()
plt.close()
```

Signups by Time of Day

```python
df['Time_of_Day'].value_counts()
```

```python
plt.figure()
sns.countplot(x='Time_of_Day', data=df, color='gold')
plt.title("Signups by Time of Day")
plt.show()
plt.close()
```

Applications

```python
df['Days_to_Apply'].describe()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure()
sns.histplot(df['Days_to_Apply'], bins=30, color='teal')
plt.title("Distribution of Days to Apply")
plt.show()
```

Engagement

```python
df['Engagement_Score'].describe()
```

```python
plt.figure()
sns.histplot(df['Engagement_Score'], bins=30, color='coral')
plt.title("Distribution of Engagement Score")
plt.show()
```

Status Distribution

```python
df['Status Description'].value_counts()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(y='Status Description', data=df, order=df['Status Description'].value_counts().index, palette='Set3', hue='Status Description', legend=False)
plt.title("Learner Status Distribution")
plt.show()
```
