```python
import pandas as pd
```

```python
df = pd.read_csv('Cleaned_Preprocessed_Dataset_Week1.csv')
```

**Descriptive Statistics**

```python
df.describe()
```

```python
df['Completed'] = df['Status Description'] == 'Applied'
df.groupby('Completed') [['Engagement_Score', 'Days_to_Apply']].describe()
```

**Engagement Score Distribution**

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,6))
sns.histplot(data=df, x='Engagement_Score', hue = 'Completed', bins=20, kde=True)
plt.title("Engagement Score Distribution")
plt.show()
```

**Engagement vs Completion**

```python
plt.figure(figsize=(8,5))
sns.boxplot(hue='Completed', y='Engagement_Score', data=df, palette='Set2', legend=False)
plt.title("Engagement Score by Completion Status")
plt.show()
```

**Days to Apply Distribution**

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='Days_to_Apply', hue='Completed', bins=20, kde=True)
plt.title("Days to Apply Distribution")
plt.show()
```
